```python
import librosa 
import torch
from transformers import HubertForCTC, AutoProcessor
```

```python
AUDIO_FILE ="/Users/dhaneshkumarkapadia/Desktop/swinTransformer/hubert/harvard.wav"
```

```python
# Load the model and tokenizer
processor = AutoProcessor.from_pretrained("facebook/hubert-large-ls960-ft")
model = HubertForCTC.from_pretrained("facebook/hubert-large-ls960-ft")
```

```python
speech, rate =  librosa.load(AUDIO_FILE, sr=16000)

inputs = processor(speech, sampling_rate= rate,return_tensors="pt")

#Model logits
with torch.no_grad():
    logits = model(inputs.input_values).logits
predicted_ids = torch.argmax(logits, dim=-1)
```

```python
transcription = processor.batch_decode(predicted_ids)
print(transcription)
```
